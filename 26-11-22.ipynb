# GC

## Problem
The GC-content of a DNA string is given by the percentage of symbols in the string that are 'C' or 'G'. For example, the GC-content of "AGCTATAG" is 37.5%. Note that the reverse complement of any DNA string has the same GC-content.

DNA strings must be labeled when they are consolidated into a database. A commonly used method of string labeling is called FASTA format. In this format, the string is introduced by a line that begins with '>', followed by some labeling information. Subsequent lines contain the string itself; the first line to begin with '>' indicates the label of the next string.

In Rosalind's implementation, a string in FASTA format will be labeled by the ID "Rosalind_xxxx", where "xxxx" denotes a four-digit code between 0000 and 9999.

### Given: 
At most 10 DNA strings in FASTA format (of length at most 1 kbp each).

### Return: 
The ID of the string having the highest GC-content, followed by the GC-content of that string. Rosalind allows for a default error of 0.001 in all decimal answers unless otherwise stated; please see the note on absolute error below.

### Sample Dataset
Rosalind_6404
CCTGCGGAAGATCGGCACTAGAATAGCCAGAACCGTTTCTCTGAGGCTTCCGGCCTTCCC
TCCCACTAATAATTCTGAGG

Rosalind_5959

ATATCCATTTGTCAGCAGACACGC

Rosalind_0808
CCACCCTCGTGGTATGGCTAGGCATTCAGGAACCGGAGAACGCTTCAGACCAGCCCGGAC
TGGGAACCTGCGGGCAGTAGGTGGAAT

# Sample Output
Rosalind_0808

file = open("rosalind_gc.txt", "r")

sequences = {}

for line in file:

    line = line.strip()

    if line.startswith(">"):
        sequence_id = line[1:]
        sequences[sequence_id] = ""

    else:
        sequences[sequence_id] += line

file.close()

print(sequences)

```python
def gc(dna):
    no_C = dna.count("C")
    no_G = dna.count("G")
    length = len(dna)
    return (no_C + no_G) / length
```

```python
sequences = {}

with open("./rosalind_gc.txt", "r") as rosalind:
    for line in rosalind:
        # figure out if it's a header or a sequence
        if line[0] == ">":
            # is a header!
            last_id = line[1:].rstrip()
            sequences[last_id] = ""
        else: # this is a sequence line!
            sequences[last_id] += line.rstrip()
```

```python
gc_content = {}
for seq_id, seq in sequences.items():
    gc_content[seq_id] = gc(seq)
```

```python
max_gc = max(gc_content.values())

for seq_id in gc_content:
    if gc_content[seq_id] == max_gc:
        print(seq_id)
        print(round(max_gc*100, 6))
```
